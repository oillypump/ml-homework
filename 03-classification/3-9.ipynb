```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

sigmoid(0), sigmoid(2), sigmoid(-2), sigmoid(5)
```

```python
import matplotlib.pyplot as plt

z = np.linspace(-7, 7, 51)
plt.plot(z, sigmoid(z))
plt.axhline(0.5, color='gray', linestyle='--')
plt.xlabel('z (skor)')
plt.ylabel('sigmoid(z)')
```
