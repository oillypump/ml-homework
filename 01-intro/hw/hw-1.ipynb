### Q6 Median value of horsepower || Yes, it decreased

```python
df.horsepower.median()
```

```python
df = pd.read_csv("car_fuel_efficiency_2026.csv")
df["horsepower"].isnull().sum()
```

```python
median_before = df["horsepower"].median()
mode_value = df["horsepower"].mode()[0]

median_before, mode_value
```

```python
df["horsepower"] = df["horsepower"].fillna(mode_value)
median_after = df["horsepower"].median()

median_before, median_after
```

### Q7. Sum of weights |||  0.36919696904925486


```python
df = pd.read_csv("car_fuel_efficiency_2026.csv")
```

```python
df["origin"].unique()
```

```python
df.isnull().sum()
```

```python
df_asia = df[df["origin"] == "Asia"]
df_asia.isnull().sum()
```

```python
df_sel = df_asia[["vehicle_weight", "model_year"]]
df_sel
```

```python
df_7 = df_sel.head(7)
df_7
```

```python
X = df_7.to_numpy()
X.shape
```

```python
XTX = X.T.dot(X)
XTX.shape    # harus (2, 2)
```

```python
import numpy as np
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv.dot(XTX)       # cek: harusnya mendekati matriks identitas
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y.shape      # (7,)
```

```python
w = XTX_inv.dot(X.T).dot(y)
w.shape
```

```python
w.sum()
```
