## multiplication

### vector-vector multiplication

```python
v.shape[0]
```

```python
u.shape[0]
```

```python
print(f"u = \n{u}")
print(f"v = \n{v}")
```

```python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0], "u and v must be the same length"

    n = u.shape[0]
    result = 0

    for i in range(n):
        # result += u[i] * v[i]
        result = result + u[i] * v[i]
    return result
```

```python
vector_vector_multiplication(u, v)
```

```python
u.dot(v)
```

### matrix-vector multiplication

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
U
```

```python
print(f"U = \n{U}")
print(f" \nU.shape[1] = {U.shape[1]}")
print(f" \nU.shape = {U.shape}")
print("\n")
print("---------------------")
print("\n")
print(f"v = \n{v}") 
print(f" \nv.shape[0] = {v.shape[0]}")
print(f" \nv.shape = {v.shape}")
```

```python
U[2, :]
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0], "U and v must be compatible for multiplication"

    num_rows = U.shape[0]
    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i, :], v)
        ## U[0, :] * v 
        ## U[1, :] * v
        ## U[2, :] * v
    return result   
```

```python
matrix_vector_multiplication(U, v)
```

```python
U.dot(v)
```

## matrix-matrix multiplication

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1], 
    [0, 2, 1],
    [2, 1, 0],
])
```

```python
print(f"U = \n{U}")
print(f"U.shape = {U.shape}")
print(f"Number of rows in U: {U.shape[0]}")
print(f"Number of columns in U: {U.shape[1]}")
print(f"U[2, :] = {U[2, :]}")
print(":---------------------")
print(f"V.shape = {V.shape}")
print(f"Number of rows in V: {V.shape[0]}")
print(f"Number of columns in V: {V.shape[1]}")
print(f"V = \n{V}")
print(":---------------------")
# print(f""assert U.shape[1] == V.shape[0])
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0], "U and V must be compatible for multiplication"

    num_rows = U.shape[0]
    num_cols = V.shape[1]
    result = np.zeros((num_rows, num_cols))

    # for i in range(num_cols):
        # vi = V[:, i]
        # Uvi = matrix_vector_multiplication(U, vi)
        # result[:, i] = Uvi
    for i in range(num_rows):
        for j in range(num_cols):
            result[i, j] = vector_vector_multiplication(U[i, :], V[:, j])
    return result
```

```python
matrix_matrix_multiplication(U, V)
```

```python
U.dot(V)
```
